```python
import torch
```

```python
# Input tensor
tensor = torch.tensor([1.0, -0.5, 3.2, -2.8, 0.0], dtype=torch.float32)

scale = 0.025
zero_point = 0
```

```python
def quantize_tensor(tensor, scale, zero_point):
    """
    Quantize a float tensor to INT8 .
    """
    q = torch.round(tensor / scale) + zero_point
    q = torch.clamp(q, -128, 127)
    return q.to(torch.int8)


def dequantize_tensor(quantized_tensor, scale, zero_point):
    """
    Dequantize an INT8 tensor back to float
    """
    return (quantized_tensor.to(torch.float32) - zero_point) * scale
```

```python
# Quantize
quantized_tensor = quantize_tensor(tensor, scale, zero_point)

# Dequantize
dequantized_tensor = dequantize_tensor(quantized_tensor, scale, zero_point)

# Error metrics
absolute_error = torch.abs(tensor - dequantized_tensor)
mae = torch.mean(absolute_error)
max_error = torch.max(absolute_error)
```

```python
print("Original tensor:")
print(tensor)

print("\nQuantized tensor (INT8):")
print(quantized_tensor)

print("\nDequantized tensor (float):")
print(dequantized_tensor)

print("\nAbsolute error per element:")
print(absolute_error)

print("\nMean Absolute Error (MAE):")
print(mae.item())

print("\nMaximum Absolute Error:")
print(max_error.item())
```
